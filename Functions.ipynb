```python
x=0
y=5
while x<y:
    print("x current value, x")
    x=x+1
    print("updated x value", x)
```

```python
def fun22(x,y):
    while x<y:
        print ("x current value", x)
        x=x+1
        print("updated x value",x)
```

```python
fun22(5,15)
```
